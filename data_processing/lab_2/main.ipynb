```python
import pandas as pd
import numpy as np
```

```python
recipes = pd.read_csv('recipes_short.csv', index_col='id')
reviews = pd.read_csv('reviews_short.csv')
```

```python
print("=== Recipes ===")
print(f"Строк: {recipes.shape[0]}, Столбцов: {recipes.shape[1]}")
print("Типы данных:")
print(recipes.dtypes)

print("\n=== Reviews ===")
print(f"Строк: {reviews.shape[0]}, Столбцов: {reviews.shape[1]}")
print("Типы данных:")
print(reviews.dtypes)
```

```python
def analyze_missing(df, name):
    nulls = df.isna().sum()
    total = df.shape[0]
    print(f"\nПропуски в {name}:")
    print(nulls[nulls > 0])
    print(f"Доля строк с пропусками: {nulls.sum() / total:.2%}")

analyze_missing(recipes, 'recipes')
analyze_missing(reviews, 'reviews')
```

```python
print("Средние значения для рецептов:")
print(recipes[['minutes', 'n_steps', 'n_ingredients']].mean())

# Для отзывов (reviews)
print("\nСредние значения для отзывов:")
print(reviews['rating'].mean())
```

```python
print("\n10 случайных названий рецептов:")
print(recipes['name'].sample(10, random_state=42))
```

```python
print("\nПервые 3 строки Reviews со старым индексом:")
print(reviews.head(3))
reviews.reset_index(drop=True, inplace=True)
print("\nПервые 3 строки Reviews с новым индексом:")
print(reviews.head(3))
```

```python
filtered_recipes = recipes[
    (recipes['minutes'] <= 20) & 
    (recipes['n_ingredients'] <= 5)
]
print("\nФильтрованные рецепты:")
print(f"Строк: {filtered_recipes.shape[0]}, Столбцов: {filtered_recipes.shape[1]}")
print(filtered_recipes.dtypes)
```

```python
recipes = pd.read_csv('recipes_short.csv', 
                     index_col='id', 
                     parse_dates=['submitted'])
```

```python
old_recipes = recipes[recipes['submitted'].dt.year <= 2010]
print("\nРецепты до 2010 года:")
print(old_recipes.head())
```

```python
recipes['description_length'] = recipes['description'].str.len()
print("\nДобавлена длина описания:")
print(recipes[['description', 'description_length']].head())
```

```python
recipes['name'] = recipes['name'].str.title()
print("\nОтформатированные названия:")
print(recipes['name'].head())
```

```python
recipes['name_word_count'] = recipes['name'].str.split().str.len()
print("\nКоличество слов в названии:")
print(recipes[['name', 'name_word_count']].head())
```

```python
contributor_stats = recipes['contributor_id'].value_counts()
top_contributor = contributor_stats.idxmax()
print(f"Участник с максимальным количеством рецептов (ID: {top_contributor}), количество: {contributor_stats.max()}")
```

```python
recipe_ratings = reviews.groupby('recipe_id')['rating'].mean()
recipes_without_reviews = recipes.index[~recipes.index.isin(reviews['recipe_id'])].size
print(f"\nРецептов без отзывов: {recipes_without_reviews}")
```

```python
recipes['year'] = pd.to_datetime(recipes['submitted']).dt.year
recipes_by_year = recipes['year'].value_counts().sort_index()
print("\nРецепты по годам:")
print(recipes_by_year)
```

```python
merged_inner = pd.merge(
    recipes.reset_index()[['id', 'name']],
    reviews[['recipe_id', 'user_id', 'rating']], 
    left_on='id',
    right_on='recipe_id',
    how='inner'
)[['id', 'name', 'user_id', 'rating']]

print("Рецепты с оценками (первые 5 строк):")
print(merged_inner.head())

no_review_id = recipes.index.difference(reviews['recipe_id']).values[0]
no_review_name = recipes.loc[no_review_id, 'name']

print(f"\nПроверка: рецепт без отзывов (ID: {no_review_id}, Название: '{no_review_name}')")
print("Поиск этого рецепта в объединенной таблице:")
print(merged_inner[merged_inner['id'] == no_review_id])
```

```python
review_counts = reviews.groupby('recipe_id').size().reset_index(name='review_count')
merged_left = pd.merge(
    recipes[['name']],
    review_counts,
    left_index=True,
    right_on='recipe_id',
    how='left'
).fillna({'review_count': 0})
print("\nРецепт без отзывов (пример):")
print(merged_left[merged_left['review_count'] == 0].head(1))
```

```python
reviews['year'] = pd.to_datetime(reviews['date']).dt.year
worst_year = reviews.groupby('year')['rating'].mean().idxmin()
print(f"\nГод с наименьшим средним рейтингом: {worst_year}")
```

```python
sorted_recipes = recipes.sort_values('name_word_count', ascending=False)
sorted_recipes.to_csv('processed_recipes.csv', index=True)
print("\nФайл processed_recipes.csv сохранен")
```

```python
with pd.ExcelWriter('results.xlsx') as writer:
    merged_inner.to_excel(writer, sheet_name='Рецепты с оценками', index=False)
    merged_left.to_excel(writer, sheet_name='Количество отзывов по рецептам', index=False)
print("Файл results.xlsx сохранен")
```
