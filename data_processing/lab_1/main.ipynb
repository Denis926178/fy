```python
import numpy as np
```

```python
data = np.loadtxt('minutes_n_ingredients.csv', delimiter=',', skiprows=1, dtype=np.int32)
print("Первые 5 строк массива:\n", data[:5], sep='')
```

```python
stats = [
    ('Среднее', np.mean),
    ('Минимум', np.min),
    ('Максимум', np.max),
    ('Медиана', np.median)
]

for name, func in stats:
    res = func(data[:, 1:], axis=0)
    print(f"{name}: время = {res[0]:.2f}, ингредиенты = {res[1]:.2f}")
```

```python
q75 = np.quantile(data[:, 1], 0.75)
data[:, 1] = np.where(data[:, 1] > q75, q75, data[:, 1])
```

```python
print(f"Рецептов с нулевой продолжительностью: {np.sum(data[:, 1] == 0)}")
data[:, 1] = np.where(data[:, 1] == 0, 1, data[:, 1])
```

```python
print(f"Уникальных рецептов: {len(np.unique(data[:, 0]))}")
```

```python
unique_ingredients, counts = np.unique(data[:, 2], return_counts=True)
print(f"Уникальные значения ингредиентов\n: {unique_ingredients}")
print(f"Количество значений:\n {counts}")
```

```python
filtered_data = data[data[:, 2] <= 5]
```

```python
print(f"Максимальное значение: {(data[:, 2] / data[:, 1]).max():.2f}")
```

```python
print(f"Среднее ингредиентов для топ-100: {(data[np.argsort(-data[:, 1])][:100])[:, 2].mean():.2f}")
```

```python
random_indices = np.random.choice(len(data), size=10, replace=False)
print(f"Случайные рецепты:\n{(data[random_indices])}")
```

```python
print(f"Процент рецептов: {np.sum(data[:, 2] < data[:, 2].mean()) / len(data) * 100:.2f}%")
```

```python
is_simple = (data[:, 1] <= 20) & (data[:, 2] <= 5)
data_with_flag = np.column_stack((data, is_simple.astype(int)))
print(f"Процент простых рецептов: {is_simple.mean() * 100:.2f}%")
```

```python
mask_short = data[:, 1] < 10
mask_standard = (data[:, 1] > 10) & (data[:, 1] < 20)
mask_long = data[:, 1] >= 20

groups = [
    data[mask_short],
    data[mask_standard],
    data[mask_long]
]

min_size = min(len(g) for g in groups)
arr3d = np.array([g[:min_size] for g in groups])
print("Форма 3D массива:", arr3d.shape)
```
