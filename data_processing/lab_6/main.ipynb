```python
recipes = pd.read_csv('recipes_sample.csv')

sample = recipes[['id', 'minutes']].dropna().sample(5)

len_first = 0
len_second = 0

max_id_len = max(sample['id'].astype(str).apply(len))
max_minutes_len = max(sample['minutes'].astype(str).apply(len))

len_first = max(max_id_len, len("id")) + 8
len_second = max(max_minutes_len, len("minutes")) + 8

header = f"|{'id':^{len_first}}|{'minutes':^{len_second}}|"
separator = f"|{'-'*(len_first)}|{'-'*(len_second)}|"

print(f"|{'id':^{len_first}}|{'minutes':^{len_second}}|")
print(separator)
for _, row in sample.iterrows():
    print(f"|{str(row['id']):^{len_first}}|{str(row['minutes']):^{len_second}}|")
```

```python
def show_info(recipe_id):
    recipe = recipes[recipes['id'] == recipe_id].iloc[0]

    tree = ET.parse('steps_sample.xml')
    root = tree.getroot()

    steps = []
    for recipe_elem in root.findall('recipe'):
        elem_id = recipe_elem.findtext('id')

        if elem_id is not None and elem_id.strip().isdigit():
            current_id = int(elem_id.strip())
            if current_id == recipe_id:
                steps = [step.text.strip() for step in recipe_elem.findall('steps/step') if step.text]
                break

    title = recipe['name'].title()
    steps_str = '\n'.join([f"    {i+1}. {step}" for i, step in enumerate(steps)])
    info = dedent(f"""
    "{title}"

{steps_str}

    ----------
    Автор: {recipe['contributor_id']}
    Среднее время приготовления: {recipe['minutes']} минут
    """)
    
    return info

print(show_info(170895))
```
